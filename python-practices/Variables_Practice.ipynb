## 3. Variables can store different types of values

Python-la common data types:

| Data type | Meaning | Example |
|---|---|---|
| `str` | Text | `"Python"` |
| `int` | Whole number | `25` |
| `float` | Decimal number | `5.75` |
| `bool` | True or False | `True` |

Example:

```python
course = "Python"       # str
lessons_completed = 3  # int
completion = 25.5      # float
is_learning = True    # bool

print(course)
print(lessons_completed)
print(completion)
print(is_learning)
```

## 5. Variable naming rules

Python variable names-ku sila rules irukku:

- Letters (`name`), numbers (`age2`), underscore (`my_age`) use pannalaam.
- Variable name number-la start aaga koodadhu. `2age` **invalid**.
- Spaces use panna koodadhu. `my age` **invalid**. Instead `my_age` use pannu.
- Python case-sensitive: `age` and `Age` rendu different variables.
- `if`, `for`, `class` maadhiri Python keywords-a variable name-a use panna koodadhu.
- Read panna easy-a irukkura descriptive names choose pannu.

Python-la usual style `snake_case`: `student_name`, `total_marks`.
